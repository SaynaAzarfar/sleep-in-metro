```python
import math, random
from typing import List, Tuple, Dict

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
import torchvision
from torchvision import transforms

import matplotlib.pyplot as plt

def set_seed(seed: int = 42):
    random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device

```

```python
class MyNNModel(nn.Module):
    """
    Fully-connected MLP for 28x28 images (flattened to 784).
    Depth = 8 hidden layers, width = 16, ReLU. Output = 10 classes.
    """
    def __init__(self, depth: int = 8, width: int = 16, in_dim: int = 784, out_dim: int = 10):
        super().__init__()
        assert depth >= 1, "Depth must be >= 1"

        layers = [nn.Linear(in_dim, width)]
        for _ in range(depth - 1):
            layers.append(nn.Linear(width, width))
        self.hidden = nn.ModuleList(layers)
        self.out = nn.Linear(width, out_dim)

    def forward_with_activations(self, x):
        """
        Forward that returns:
          - logits
          - list of post-activation tensors per hidden layer (for retain_grad)
        """
        x = x.view(x.size(0), -1)
        activations = []
        for layer in self.hidden:
            x = layer(x)
            x = F.relu(x)
            activations.append(x)
        logits = self.out(x)
        return logits, activations

    def forward(self, x):
        x = x.view(x.size(0), -1)
        for layer in self.hidden:
            x = F.relu(layer(x))
        return self.out(x)

```

```python
def init_uniform(model: nn.Module, d: float):
    """Linear weights ~ U(-d, d); biases = 0."""
    for m in model.modules():
        if isinstance(m, nn.Linear):
            nn.init.uniform_(m.weight, a=-d, b=d)
            nn.init.constant_(m.bias, 0.0)

def init_xavier_uniform(model: nn.Module):
    """Xavier uniform: U(-sqrt(6/(fan_in+fan_out)), +sqrt(...)); biases = 0."""
    for m in model.modules():
        if isinstance(m, nn.Linear):
            nn.init.xavier_uniform_(m.weight)
            nn.init.constant_(m.bias, 0.0)

def apply_init(model: nn.Module, mode: str):
    """
    mode ∈ {"0.01", "0.1", "1.0", "xavier"}
    """
    if mode == "xavier":
        init_xavier_uniform(model)
    elif mode in {"0.01", "0.1", "1.0"}:
        init_uniform(model, float(mode))
    else:
        raise ValueError(f"Unknown init mode: {mode!r}")

```

```python
def get_mnist_loader(batch_size: int = 128) -> DataLoader:
    # Standard MNIST normalization (mean=0.1307, std=0.3081)
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.1307,), (0.3081,)),
    ])
    ds = torchvision.datasets.MNIST(root="./data", train=True, download=True, transform=transform)
    # drop_last=True guarantees a batch of exactly 128
    return DataLoader(ds, batch_size=batch_size, shuffle=True, num_workers=2, drop_last=True)

def get_fashion_loaders(batch_size: int = 128) -> Tuple[DataLoader, DataLoader]:
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,)),
    ])
    train_set = torchvision.datasets.FashionMNIST(root="./data", train=True,  download=True, transform=transform)
    test_set  = torchvision.datasets.FashionMNIST(root="./data", train=False, download=True, transform=transform)
    train_loader = DataLoader(train_set, batch_size=batch_size, shuffle=True,  num_workers=2)
    test_loader  = DataLoader(test_set,  batch_size=batch_size, shuffle=False, num_workers=2)
    return train_loader, test_loader
```

```python
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

@torch.no_grad()
def _l2(x: torch.Tensor) -> float:
    return float(torch.linalg.norm(x).cpu())

def compute_grad_norms_for_init(init_mode: str, batch):
    """
    - Build model, apply `init_mode`
    - Forward a single batch, retain_grad() on each post-activation 'a'
    - Backward CE once
    - Return L2 norms of ∥∂L/∂a∥ for the 8 layers
    """
    x, y = batch[0].to(device), batch[1].to(device)
    model = MyNNModel(depth=8, width=16, in_dim=784, out_dim=10).to(device)
    apply_init(model, init_mode)

    logits, activations = model.forward_with_activations(x)
    for a in activations:
        a.retain_grad()

    loss = nn.CrossEntropyLoss()(logits, y)
    model.zero_grad(set_to_none=True)
    loss.backward()

    norms = [_l2(a.grad) for a in activations]
    return norms


# -----------------------------
# Get MNIST minibatch
# -----------------------------
mnist_loader = get_mnist_loader(batch_size=128)
mnist_batch = next(iter(mnist_loader))

init_modes = ["0.01", "0.1", "1.0", "xavier"]
grad_norms = {mode: compute_grad_norms_for_init(mode, mnist_batch) for mode in init_modes}


# -----------------------------
# Plot: 4 separate graphs (one per init)
# -----------------------------
layers_x = list(range(1, 9))

for mode in init_modes:
    plt.figure()
    plt.plot(layers_x, grad_norms[mode], marker='o')
    plt.xlabel("Layer (post-activation, 1 = first hidden)")
    plt.ylabel("||∂L/∂a|| (L2 norm)")
    plt.title(f"MNIST — Gradient norms per layer\nInitialization = {mode}")
    plt.grid(True)
    plt.show()
```

```python
def accuracy_from_logits(logits: torch.Tensor, targets: torch.Tensor) -> float:
    preds = logits.argmax(dim=1)
    correct = (preds == targets).sum().item()
    return correct / targets.size(0)
def train_eval_5_epochs(init_mode: str, lr: float = 0.01, batch_size: int = 128):
    """
    Train 5 epochs (SGD, lr=0.01, batch=128), record (train_acc, test_acc) per epoch.
    Returns: (list_train_acc, list_test_acc)
    """
    train_loader, test_loader = get_fashion_loaders(batch_size=batch_size)
    model = MyNNModel(depth=8, width=16, in_dim=784, out_dim=10).to(device)
    apply_init(model, init_mode)

    optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    train_accs, test_accs = [], []
    for epoch in range(1, 6):
        # ---- train ----
        model.train()
        correct, total = 0, 0
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad(set_to_none=True)
            logits = model(x)
            loss = criterion(logits, y)
            loss.backward()
            optimizer.step()

            correct += (logits.argmax(dim=1) == y).sum().item()
            total += y.size(0)
        train_acc = correct / total
        train_accs.append(train_acc)

        # ---- eval ----
        model.eval()
        correct, total = 0, 0
        with torch.no_grad():
            for x, y in test_loader:
                x, y = x.to(device), y.to(device)
                logits = model(x)
                correct += (logits.argmax(dim=1) == y).sum().item()
                total += y.size(0)
        test_acc = correct / total
        test_accs.append(test_acc)

        print(f"[{init_mode}] Epoch {epoch}/5 | Train acc {train_acc*100:.2f}% | Test acc {test_acc*100:.2f}%")

    return train_accs, test_accs

results_train, results_test = {}, {}
for mode in init_modes:
    tr, te = train_eval_5_epochs(mode, lr=0.01, batch_size=128)
    results_train[mode] = tr
    results_test[mode] = te

```

```python
import matplotlib.pyplot as plt

epochs = [1, 2, 3, 4, 5]

# Loop over each initialization mode and make one separate graph
for mode in init_modes:
    plt.figure()
    plt.plot(epochs, results_train[mode], marker='o')
    plt.xlabel("Epoch")
    plt.ylabel("Training accuracy")
    plt.title(f"FashionMNIST — Training accuracy vs epochs\n(SGD, lr=0.01, batch=128, init={mode})")
    plt.grid(True)
    plt.show()

```

```python
# Xavier-only, 3 optimizers: SGD, RMSprop, Adam (lr=0.01), batch=128, 5 epochs
def train_eval_5_epochs_with_optimizer(optimizer_name: str, lr: float = 0.01, batch_size: int = 128):
    """
    Xavier init, CrossEntropyLoss, train 5 epochs on FashionMNIST with the chosen optimizer.
    Returns (train_accs, test_accs).
    """
    train_loader, test_loader = get_fashion_loaders(batch_size=batch_size)

    model = MyNNModel(depth=8, width=16, in_dim=784, out_dim=10).to(device)
    apply_init(model, "xavier")  # <— Xavier only

    optimizer_name = optimizer_name.lower()
    if optimizer_name == "sgd":
        optimizer = torch.optim.SGD(model.parameters(), lr=lr)
    elif optimizer_name == "rmsprop":
        optimizer = torch.optim.RMSprop(model.parameters(), lr=lr, alpha=0.99)
    elif optimizer_name == "adam":
        optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    else:
        raise ValueError(f"Unknown optimizer: {optimizer_name}")

    criterion = nn.CrossEntropyLoss()

    train_accs, test_accs = [], []
    for epoch in range(1, 6):
        # ---- train ----
        model.train()
        correct, total = 0, 0
        for x, y in train_loader:
            x, y = x.to(device), y.to(device)
            optimizer.zero_grad(set_to_none=True)
            logits = model(x)
            loss = criterion(logits, y)
            loss.backward()
            optimizer.step()

            correct += (logits.argmax(dim=1) == y).sum().item()
            total += y.size(0)
        train_acc = correct / total
        train_accs.append(train_acc)

        # ---- eval ----
        model.eval()
        correct, total = 0, 0
        with torch.no_grad():
            for x, y in test_loader:
                x, y = x.to(device), y.to(device)
                logits = model(x)
                correct += (logits.argmax(dim=1) == y).sum().item()
                total += y.size(0)
        test_acc = correct / total
        test_accs.append(test_acc)

        print(f"[{optimizer_name.upper()} | Xavier] Epoch {epoch}/5 | "
              f"Train {train_acc*100:.2f}% | Test {test_acc*100:.2f}%")

    return train_accs, test_accs

# Run all three optimizers
opt_list = ["sgd", "rmsprop", "adam"]
xavier_results_train = {}
xavier_results_test  = {}
for opt in opt_list:
    tr, te = train_eval_5_epochs_with_optimizer(opt, lr=0.01, batch_size=128)
    xavier_results_train[opt] = tr
    xavier_results_test[opt]  = te

```

```python
# Training accuracy curves (one plot, three curves)
plt.figure()
epochs = [1, 2, 3, 4, 5]
for opt in opt_list:
    plt.plot(epochs, xavier_results_train[opt], marker='o', label=opt.upper())
plt.xlabel("Epoch")
plt.ylabel("Training accuracy")
plt.title("FashionMNIST — Training accuracy (Xavier init, lr=0.01, batch=128)")
plt.legend()
plt.show()

# Testing accuracy curves (one plot, three curves)
plt.figure()
for opt in opt_list:
    plt.plot(epochs, xavier_results_test[opt], marker='o', label=opt.upper())
plt.xlabel("Epoch")
plt.ylabel("Testing accuracy")
plt.title("FashionMNIST — Testing accuracy (Xavier init, lr=0.01, batch=128)")
plt.legend()
plt.show()
```
